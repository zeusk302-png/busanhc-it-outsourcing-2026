# 2단계 · list와 dict로 상품 조회
상품 목록과 상품별 가격·재고를 조회합니다. 시작본은 product_info의 TODO 한 곳을 채우세요.
인터넷/API/실제 고객 자료를 사용하지 않는 독립 실습입니다.


```python
PRODUCTS = {
    "키링": {"price": 4000, "stock": 3},
    "스티커": {"price": 1000, "stock": 6},
    "노트": {"price": 3000, "stock": 4},
}

PRODUCT_NAMES = ["키링", "스티커", "노트"]


def product_info(name):
    if name not in PRODUCTS:
        return None
    # TODO 2: PRODUCTS에서 단가와 재고를 찾아 위와 같은 사전으로 반환하세요.
    raise NotImplementedError("TODO 2: 상품 정보 조회")


def main():
    print("상품 목록:", PRODUCT_NAMES)
    name = input("확인할 상품: ").strip()
    info = product_info(name)
    if info is None:
        print("목록에 없는 상품입니다.")
    else:
        print(info["name"], "단가:", info["price"], "남은 수량:", info["stock"])

```
